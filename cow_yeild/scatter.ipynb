```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = [4,6,10,12]
y = [3,5.5,6.5,9]

yc = []

a0 = 0.8
a1 = 0.65

# a0 = 0.8 a1 = 0.65 calculated before for regression line

# to plot need to calculate many values of y from x OR just use set of x values to calculate corresponding y values

for i in x:
    yc.append(a0+a1*i)
    
```

```python
plt.plot(x,y,marker='o',label="x vs y")
plt.plot(x,yc,label="regression line")
plt.xlabel("cow feed")
plt.ylabel("cow yeild")
plt.legend()
plt.grid(True)
plt.title("cow feed vs yeild")
plt.show()
```
