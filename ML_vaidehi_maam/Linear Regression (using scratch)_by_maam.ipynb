```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Regular Method

```python
def regular_method(x, y): 
    n = len(x) 
    x_mean = np.mean(x) 
    y_mean = np.mean(y)
    
    print(f"Mean of x: {x_mean}") 
    print(f"Mean of y: {y_mean}")
    
    # b1 (slope) 
    numerator = np.sum((x - x_mean) * (y - y_mean)) 
    denominator = np.sum((x - x_mean) ** 2) 
    a1 = numerator / denominator
    
    # b0 (intercept)
    a0 = y_mean - a1 * x_mean
    print(a0)
    print(a1)
    
    y_pred = a0 + a1 * x 
    print(f"y_pred: {y_pred}") 
    error = y - y_pred 
    print(f"error: {error}")
    
    plt.scatter(x, y, color="blue", label="Actual Data") 
    plt.plot(x, y_pred, color="red", label="Regression Line") 
    plt.xlabel("x") plt.ylabel("y") 
    plt.title("LinearRegression(Regular Method)") 
    plt.legend() 
    plt.grid() 
    plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 1, 2, 2, 4])

print(f"Birth Weight Data:")
print(f"x:{x}\ny:{y}")

regular_method(x,y)
```
