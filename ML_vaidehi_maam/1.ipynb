```python
# why np.mean and not pd.mean
# pandas.mean for dataframe like excel sheet
# and numpy.mean for normal python array

def regular_method(x,y):
    n = len(x) # no of observations
    xbar = np.mean(x)
    ybar = np.mean(y)
# slope
    temp = 0
    temp1 = 0
    for i in range(0,n):
        temp = temp + (x[i]-xbar)*(y[i]-ybar)
        temp1 = temp1 + (x[i]-xbar)**2


    a1 = temp / temp1

# intercept a0

    a0 = ybar - (a1 * xbar)

    ycap = []
    temp = 0

    for i in range(0,n):
        yc = a0 + a1*x[i]
        ycap.append(yc)

        ei_2 = (y[i] - yc)**2

        temp = temp + ei_2
        

    
    sse = temp

    return (a0,a1,ycap,sse)
        
        
```

```python
height = np.array([1,2,3,4,5])
weight = np.array([1,1,2,2,4])
```

```python
a0_a1 = regular_method(height,weight)
print(round(a0_a1[0],4))
print(round(a0_a1[1],4))

ycap = a0_a1[2]
sse = a0_a1[3]
print(round(sse,4))
```

```python
# plt.scatter(height,weight)
plt.scatter(height,weight,color="red")
plt.plot(height,ycap)
plt.xlabel("height")
plt.ylabel("weight")
plt.grid(True)
plt.show()
```
